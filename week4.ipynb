why do we need activation functions?
suppose neuron computes 
z = w1x1 + w2x2 + .... + b
this is just a linear eqn

now imagine stacking layers:
z1 = w1x+b1
z2 = w2z1 + b2

substituting:
z2 = w2(w1x + b1) + b2
z2 = w1w2x + w2b1 + b2
still linear

so linear layer+ linear layer +linear layer  = one big linear layer

a network without activation functions cannot learn complex patterns such as:
- circles
- spirals
- faces
- speech
- handwritten digits

Activation functions introduce non-linearity, allowing neural networks to model complex relationships. 


























choosing activation function:
1. relu - regression y>=0
2. sigmoid - classification
3. linear - regression y<0 and y>=0
