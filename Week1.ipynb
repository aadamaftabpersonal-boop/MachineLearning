types of machine learning:
1. supervised machine learning:
    -   you try and find relationship between input and output
    -   given a dataset, you train your model and then use it to predict on fresh data
    -   examples:
            1. email - spam, not spam
            2. audio - text transcripts
            3. English to Spanish
            4. ad, user info - click? (0/1)
            5. image, radar_info - position of other cars
            6. image of phone - finding defects
            7. house price prediction - (regression problem) - can fit a line or a curve to the dataset as per exploring data
    -   two types: regression and classification problems
            1. cancer in women - malign or benign breast tumor - classification (0/1)
        classification algorithms predict categories - small number of possible outputs - can use two or more inputs
    -the learning algorithm has a function (hypothesis) where f(x1 | x2 | ... | xn) = yhat (yhat is called the prediction for a given x)
    fw,b (x) = wx+b - linear regression with one variable
    w, b parameters
    we start of with a linear model to understand and advance to complex models later
    cost function:
        J(w,b) = (1/2m) * sigma(1 to m) (ybar-y) ^ 2 - standard used in industry, but just the sigma part also works fine  
    goal: minimise J(w,b)
    goal of linear regression - minimise w, general goal of linear regression - minimise w, b
    for simplification, we set b=0 while minimising w
    we try to minimise/gradient descent on J(w,b), w, and b 3d graph 

    Gradient descent algorithm:
        -w = w - alpha * d(J(w, b))/dw; alpha is called learning rate
        -b = b - alpha * d(J(w, b))/dw; alpha is called learning rate
        b set to zero to make it a 2d graph of J(w) vs w
        the partial derivative is the slope, where slope negative w increases which is bad and where slope positive w decreases which is good for us
        the learning rate, alpha:
            - if too small then takes a lot of computation
            - if too large, it overshoots from the minimum and each time alpha modifies to a higher value we keep going away from the minimum, hence causing unnecessary expensive oscillations
        Gradient descent algorithm:
            - repeat till convergence{
                w = w - alpha * (1/m) [sigma(1 to m) (f(x(i))-y(i))] * x(i)
                b = b - alpha * (1/m) [sigma(1 to m) (f(x(i))-y(i))]
            }
        having the cost function vs w in 2d leads to convex optimisation wherre a 2d graph/convex function always leads to a global minimum and no local minimum
        Batch gradient descent - each step of the gradient descent uses all the training examples
        for linear regression, the cost function is convex and hence we use convex optimisation

    - dna clustering
    - there are types:
        -clustering: group similar data together
        -dimensionality reduction: compress data using fewer numbers
        -anomaly detection: find unusual points
